```r
# Load required libraries
library(readxl)
library(caret)
library(earth)
library(Metrics)
```

```r
# Load the dataset
df <- read_excel("/content/Exercise 15 Regression Trees Strength2.xlsx")

# Define features and target
X <- df[, colnames(df) != "Strength2"]
y <- df$Strength2
```

```r
# Split into training and validation sets
set.seed(42)
train_index <- createDataPartition(y, p = 0.7, list = FALSE)
X_train <- X[train_index, ]
X_val   <- X[-train_index, ]
y_train <- y[train_index]
y_val   <- y[-train_index]

# Initialize the MARS model (Earth Regressor)
mars_model <- earth(x = X_train, y = y_train)

# Predict on both train and validation sets
y_train_pred <- predict(mars_model, X_train)
y_val_pred   <- predict(mars_model, X_val)
```

```r
# Evaluate Model Performance

## R² score
r2_train <- R2(y_train_pred, y_train)
r2_val   <- R2(y_val_pred, y_val)

## RMSE (Root Mean Squared Error)
rmse_train <- rmse(y_train, y_train_pred)
rmse_val   <- rmse(y_val, y_val_pred)

## Display performance metrics
cat("Training R²:", round(r2_train, 4), "\n")
cat("Training RMSE:", round(rmse_train, 2), "\n")
cat("Validation R²:", round(r2_val, 4), "\n")
cat("Validation RMSE:", round(rmse_val, 2), "\n")
```
